```python
from google.colab import drive
drive.mount('/content/drive')
import os
P3A_DIR = '/content/drive/MyDrive/littlefig-p3a'
os.makedirs(P3A_DIR, exist_ok=True)
RESULTS_PATH = P3A_DIR + '/lowram_allocator_results.json'
LOG_PATH = P3A_DIR + '/lowram_allocator.log'
```

```python
import subprocess, json, os
cmd = ['python', '-u', 'benchmark/experiment_lowram_allocator_v1.py',
       '--iterations', '20', '--batch-size', '2', '--sequence-length', '256',
       '--results-path', RESULTS_PATH]
print('Command:', ' '.join(cmd), flush=True)
print('Durable log:', LOG_PATH, flush=True)
with open(LOG_PATH, 'w', buffering=1) as log_file:
    process = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in iter(process.stdout.readline, ''):
        print(line, end='', flush=True)
        log_file.write(line)
    process.stdout.close()
    RETURN_CODE = process.wait()
print('Return code:', RETURN_CODE, flush=True)
if os.path.exists(RESULTS_PATH):
    print(json.dumps(json.load(open(RESULTS_PATH)), indent=2), flush=True)
```
